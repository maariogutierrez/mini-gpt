# Tokenizer Test Notebook

Test the tokenizer wrapper by encoding/decoding text and inspecting token IDs.

```python
import sys
from pathlib import Path

# Add project root to path so we can import model
project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

from model.architecture.tokenizer import Tokenizer

# Initialize tokenizer
tokenizer = Tokenizer()
print(f"Tokenizer initialized with encoding: cl100k_base")
```

## 1. Load Sample Text

We'll use a paragraph of text to demonstrate encoding and decoding.

```python
text = """The quick brown fox jumps over the lazy dog. This sentence contains every letter of the 
English alphabet. Tokenization is the process of breaking down text into smaller units called tokens, 
which can be individual characters, words, or subwords. The tiktoken library uses Byte Pair Encoding 
to efficiently represent text as numerical IDs that machine learning models can process."""

print(f"Original text:\n{text}\n")
print(f"Text length: {len(text)} characters")
```

## 2. Encode Text to Token IDs

Convert the text into token IDs that the model can process.

```python
token_ids = tokenizer.encode(text)

print(f"Number of tokens: {len(token_ids)}\n")
print(f"Token IDs (first 50): {token_ids[:50]}\n")
print(f"Token IDs (last 20): {token_ids[-20:]}")
```

## 3. Decode Tokens Back to Text

Convert the token IDs back to text and verify it matches the original.

```python
decoded_text = tokenizer.decode(token_ids)

print(f"Decoded text:\n{decoded_text}\n")

# Verify the decoded text matches the original
if decoded_text == text:
    print("✓ SUCCESS: Decoded text matches the original text!")
else:
    print("✗ ERROR: Decoded text does not match the original!")
    print(f"\nOriginal length: {len(text)}")
    print(f"Decoded length: {len(decoded_text)}")
```

## 4. Vocabulary Size

Check the total number of unique tokens in the tokenizer's vocabulary.

```python
vocab_size = tokenizer.vocab_size

print(f"Vocabulary size: {vocab_size:,} unique tokens")
print(f"\nThis means the model can represent {vocab_size} different token IDs.")
```
